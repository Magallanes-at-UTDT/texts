You have some PDFs on your machine:

```python
from pathlib import Path

RAW_DIR = Path("data/raw")

pdf_files = sorted(RAW_DIR.glob("*.pdf"))

pdf_files
```

```python
# !pip install pymupdf
```

Read PDFs and save them in a dictionary:

```python
import pymupdf as fitz
```

```python
documents = {}

for pdf in pdf_files:
    doc = fitz.open(pdf)

    text = ""

    for page in doc:
        text += page.get_text()

    documents[pdf.stem] = text
```

```python
documents.keys()
```

From dict to DF:

```python
import pandas as pd

corpus = (
    pd.DataFrame.from_dict(documents,
                           orient="index",
                           columns=["text_raw"])
      .reset_index(names="party")
)
```

Adding some columns:

```python
corpus["n_characters"] = corpus["text_raw"].str.len()
```

```python
corpus["n_words"] = corpus["text_raw"].str.split().str.len()
corpus[["party", "n_characters", "n_words"]]
```

We have:

```python
corpus
```

## Save raw corpus

```python
from pathlib import Path
 
PROCESSED_DIR = Path("data/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
 
corpus.to_csv(PROCESSED_DIR / "RawCorpus.csv", index=False)
```
