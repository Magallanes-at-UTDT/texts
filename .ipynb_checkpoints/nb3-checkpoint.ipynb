```python
import pandas as pd

dtm_regex_URL = "https://github.com/Magallanes-at-UTDT/texts/raw/main/data/processed/DocumentTermMatrix_Regex_Counts.parquet"

dtm_regex = pd.read_parquet(dtm_regex_URL)
```

```python
dtm_regex.info()
```

```python
dtm_regex.index
```

calculate the total frequency of every term across the corpus:

```python
from sklearn.feature_extraction.text import TfidfTransformer

tfidf = TfidfTransformer()

dtm_regex_tfidf = pd.DataFrame(
    tfidf.fit_transform(dtm_regex).toarray(),
    index=dtm_regex.index,
    columns=dtm_regex.columns
)
```

```python
dtm_regex_tfidf.head()
```

> Raw counts measure frequency, whereas TF-IDF measures importance relative to the corpus. Consequently, TF-IDF often downweights common terms (e.g., estado, economía) while highlighting terms that distinguish one document from the others.

```python
dtm_regex_tfidf[["economía", "salud", "educación"]]
```

```python
dtm_regex[["economía", "salud", "educación"]]
```

```python
from pathlib import Path

PROCESSED_DIR = Path("data/processed")


dtm_regex_tfidf.to_parquet(
    PROCESSED_DIR / "DocumentTermMatrix_Regex_TFIDF.parquet"
)
```
