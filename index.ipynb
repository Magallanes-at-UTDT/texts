```python
documents = {}

for pdf in pdf_files:
    doc = fitz.open(pdf)

    text = ""

    for page in doc:
        text += page.get_text()

    documents[pdf.stem] = text
```

```python
documents.keys()
```

```python
for party, text in documents.items():
    print(party, len(text))
```

```python
 for pdf in pdf_files:
    doc = fitz.open(pdf)
    print(pdf.stem, len(doc))
```

```python
import pandas as pd

corpus = (
    pd.DataFrame.from_dict(documents,
                           orient="index",
                           columns=["text_raw"])
      .reset_index(names="party")
)
```

```python
corpus
```

```python
corpus["n_characters"] = corpus["text_r"].str.len()
```

```python
corpus[["party", "n_characters"]]
```

```python
corpus["n_words"] = corpus["text"].str.split().str.len()
corpus[["party", "n_characters", "n_words"]]
```

```python
from pathlib import Path

PROCESSED_DIR = Path("data/processed")
PROCESSED_DIR.mkdir(exist_ok=True)
```

```python
corpus.to_csv(PROCESSED_DIR / "corpus_raw.csv", index=False)
```
